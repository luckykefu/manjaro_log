## 混音阶段（Mixing）

顺序：增益调整 - 平衡各轨道音量 - EQ - 压缩 - 去齿音 - 混响/延迟 - 自动化

增益调整 - 平衡各轨道音量

EQ（均衡） - FabFilter Pro-Q 3

切除低频杂音（<80Hz）

调整频率平衡

压缩 - 控制动态范围

去齿音 - De-Esser（4-8kHz）

混响/延迟 - Valhalla Reverb

自动化 - 音量、效果自动化

4. 母带处理（Mastering）
   Ozone 或手动处理

EQ 整体平衡

多段压缩

限制器（-0.3dB 峰值）

目标响度：-14 LUFS（流媒体标准）

5. 导出（Export）
   WAV 24bit/48kHz（高质量）

MP3 320kbps（分发）

关键原则
从减法开始（先 EQ 切除，再增益）

少即是多（不过度处理）

休息耳朵（每小时休息 10 分钟）

参考轨（对比专业作品）
