## plugins

### Cursor 插件列表

当前已安装的 Cursor 插件：

1. **amazonwebservices.amazon-q-vscode** - Amazon Q AI 助手
2. **anysphere.cursorpyright** - Cursor Pyright（Cursor 专用 Python 类型检查）
3. **esbenp.prettier-vscode** - Prettier 代码格式化
4. **github.github-vscode-theme** - GitHub 主题
5. **ms-ceintl.vscode-language-pack-zh-hans** - 中文（简体）语言包
6. **ms-python.python** - Python 支持
7. **ms-toolsai.jupyter** - Jupyter 支持
8. **ritwickdey.liveserver** - Live Server
9. **vscode-icons-team.vscode-icons** - 文件图标


### 安装 Cursor 插件命令

```bash
# 安装单个插件
cursor --install-extension <插件ID>

# 批量安装插件
cursor --install-extension amazonwebservices.amazon-q-vscode
cursor --install-extension esbenp.prettier-vscode
cursor --install-extension github.github-vscode-theme
cursor --install-extension ms-ceintl.vscode-language-pack-zh-hans
cursor --install-extension ms-python.python
cursor --install-extension ms-toolsai.jupyter
cursor --install-extension ritwickdey.liveserver
cursor --install-extension vscode-icons-team.vscode-icons
```
