## 1. 什么是 Suno AI？

Suno AI 是一个 **人工智能音乐生成工具**，输入一段文字描述，它就能自动为你生成：

- 完整的旋律
- 伴奏（吉他、鼓、钢琴等）
- 歌词
- 真人风格的人声演唱

🎵 示例：你输入
```
一首关于夏天海边的流行歌，女声轻快演唱，有吉他和海浪声。
```
Suno 就会生成一首完整的歌曲！

## 4. 写一个好用的提示词（Prompt）

📌 **提示词 = 你对音乐的描述**

写得越清楚，AI 生成得越好！

### ✅ 提示词公式（中文版）

```
一首【音乐类型】，关于【主题】，【人声类型】演唱，
乐器包括【乐器1】、【乐器2】，节奏【快/慢/BPM】，
情绪是【开心/忧伤/梦幻】，结构：【主歌→副歌→桥段】
```

### 🌟 示例

```text
一首温暖的民谣，关于一个人在冬天看雪，男声低沉演唱，
有木吉他和风铃声，节奏缓慢，情绪安静孤独，
结构：主歌 → 副歌 → 主歌 → 副歌 → 桥段
```

```text
一首欢快的电子舞曲，庆祝生日派对，女生清脆歌声，
有合成器、鼓机和欢呼声，BPM 120，适合跳舞！
```

## 6. 试听与优化

### 🔍 如果不满意？别担心！

AI 第一次不一定完美，你可以：

- **重新生成**：点击同一提示词再试一次（每次结果不同）
- **修改提示词**：比如加上 `"更明显的副歌"` 或 `"换成女声"`
- **换风格**：比如改成 `"爵士钢琴"` 或 `"Lo-fi beat"`

### 💡 小技巧

如果你想让音乐更“抓耳”，可以加这些词：
- `"有记忆点的副歌"`
- `"渐强进入高潮"`
- `"结尾淡出"`
- `"加入环境音效"`（如鸟叫、海浪、钟声）

## 7. 下载歌曲
