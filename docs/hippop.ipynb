# Hip-Hop 音乐分类与代表作品

## 1. 🎤 Old School Hip-Hop (老派嘻哈) - 1970s-1980s

### 🎵 音乐特点
- 简单的节奏、派对氛围、DJ 打碟
- 强调韵律和节拍
- 使用黑胶唱片打碟技术

- **代表艺人**: 
  - Grandmaster Flash - "The Message" (1982)
  - Run-DMC - "Walk This Way" (1986)
  - Sugarhill Gang - "Rapper's Delight" (1979)


```python
from IPython.display import YouTubeVideo
YouTubeVideo('gYMkEMCHtJ4')  # Grandmaster Flash - The Message
```

## 2. 🔫 Gangsta Rap (匪帮说唱) - 1980s-1990s

### 🎵 音乐特点
- 描述街头生活、暴力、社会问题
- 真实反映底层社区现实
- 强烈的节奏和攻击性歌词

- **代表艺人**:
  - N.W.A - "Straight Outta Compton" (1988)
  - Tupac - "California Love" (1995)
  - The Notorious B.I.G. - "Juicy" (1994)
  - Ice Cube - "It Was a Good Day" (1992)


```python
YouTubeVideo('_JZom_gVfuw')  # Tupac - California Love
```

## 3. 🗽 East Coast Hip-Hop (东岸嘻哈)

### 🎵 音乐特点
- 复杂的歌词、爵士采样、硬核节奏
- 注重说唱技巧和文字游戏
- 纽约为中心的文化影响

- **代表艺人**:
  - Nas - "N.Y. State of Mind" (1994)
  - Wu-Tang Clan - "C.R.E.A.M." (1993)
  - Jay-Z - "Empire State of Mind" (2009)
  - A Tribe Called Quest - "Can I Kick It?" (1990)


```python
YouTubeVideo('PBwAxmrE194')  # Wu-Tang Clan - C.R.E.A.M.
```

## 4. 🌴 West Coast Hip-Hop (西岸嘻哈)

### 🎵 音乐特点
- G-Funk 风格、合成器、放克贝斯
- 慵懒的节奏和旋律
- 加州阳光文化氛围

- **代表艺人**:
  - Dr. Dre - "Nuthin' but a 'G' Thang" (1992)
  - Snoop Dogg - "Gin and Juice" (1993)
  - Kendrick Lamar - "HUMBLE." (2017)


```python
YouTubeVideo('tvTRZJ-4EyI')  # Kendrick Lamar - HUMBLE.
```

## 5. 🔥 Trap (陷阱音乐) - 2000s-Present

### 🎵 音乐特点
- 808 鼓机、Hi-Hat 快速滚动、黑暗氛围
- 重低音和电子合成器
- 南方亚特兰大起源

- **代表艺人**:
  - T.I. - "What You Know" (2006)
  - Future - "Mask Off" (2017)
  - Migos - "Bad and Boujee" (2016)
  - Travis Scott - "SICKO MODE" (2018)


```python
YouTubeVideo('xvZqHgFz51I')  # Travis Scott - SICKO MODE
```

## 6. 🧠 Conscious Hip-Hop (意识说唱)

### 🎵 音乐特点
- 社会评论、政治意识、积极信息
- 深刻的歌词内容
- 关注社会正义和平等

- **代表艺人**:
  - Common - "The Light" (2000)
  - Mos Def - "Ms. Fat Booty" (1999)
  - Kendrick Lamar - "Alright" (2015)
  - J. Cole - "Middle Child" (2019)

## 7. 🎶 Mumble Rap (含糊说唱) - 2010s

### 🎵 音乐特点
- 旋律优先、自动调音、简单歌词
- 注重音乐性而非歌词
- 大量使用 Auto-Tune

- **代表艺人**:
  - Lil Uzi Vert - "XO Tour Llif3" (2017)
  - Playboi Carti - "Magnolia" (2017)
  - Lil Yachty - "One Night" (2016)


```python
YouTubeVideo('WrsFXgQk5UI')  # Lil Uzi Vert - XO Tour Llif3
```

- **代表艺人**:
  - OutKast - "Hey Ya!" (2003)
  - Tyler, The Creator - "EARFQUAKE" (2019)
  - Kid Cudi - "Day 'n' Nite" (2008)
  - Kanye West - "Stronger" (2007)

- 融合朋克摇滚元素

- **代表艺人**:
  - Juice WRLD - "Lucid Dreams" (2018)
  - XXXTentacion - "SAD!" (2018)
  - Lil Peep - "Awful Things" (2017)
