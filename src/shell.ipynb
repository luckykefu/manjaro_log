## ln


```python
%%bash
#!/usr/bin/env bash
# 编写一个 sh 文件:
# 功能:创建连接
# 参数:
# src:元路径
# tgt:目标路径

# 强制覆盖已存在的链接
ln -sfv /data/.manjaro .manjaro
ln -sfv /home/lkf/.aws/amazonq .amazonq

# 批量创建（遍历目录）
for dir in /data/.home/*; do
    ln -sf "$dir" ~/"$(basename "$dir")"
done
```

## line to zshrc


```python
%%bash
#!/usr/bin/env bash
# 将 path 添加到 zshrc

dir="/data/projects_ING/crypto_quant"
path_line="export PATH=\"$dir:\$PATH\""
zshrc="$HOME/.zshrc"

# 防止重复添加
grep -qF "$path_line" "$zshrc" || echo "$path_line" >> "$zshrc"
echo "已添加到 $zshrc"
tail -n 3 "$zshrc"
```

```python
%%bash
#!/usr/bin/env bash
# 将 source_line
f="/data/.zshplugins/zsh-sudo/zsh-sudo.zsh"
source_line="source $f"
zshrc="$HOME/.zshrc"

# 防止重复添加
grep -qF "$source_line" "$zshrc" || echo "$source_line" >> "$zshrc"
tail -n 3 "$zshrc"
```
