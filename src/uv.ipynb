## 12. 工作区（Workspace）


```python
%%bash
# pyproject.toml 配置工作区
cat >> pyproject.toml << 'EOF'

[tool.uv.workspace]
members = ["packages/*"]
EOF
```

## 13. 命令对比

| 功能     | pip/conda              | uv                 |
| -------- | ---------------------- | ------------------ |
| 安装包   | `pip install pkg`      | `uv add pkg`       |
| 卸载包   | `pip uninstall pkg`    | `uv remove pkg`    |
| 列出包   | `pip list`             | `uv pip list`      |
| 创建环境 | `python -m venv .venv` | `uv venv`          |
| 运行脚本 | `python script.py`     | `uv run script.py` |
| 锁定依赖 | `pip freeze`           | `uv lock`          |


## 14. 完整项目示例


```python
%%bash
# 创建新项目
uv init my-project
cd my-project

# 添加依赖
uv add requests pandas
uv add --dev pytest black

# 运行代码
uv run python hello.py

# 运行测试
uv run pytest

# 格式化代码
uv run black .
```

## 15. 最佳实践

1. **使用 `uv add` 而非 `uv pip install`** - 自动更新 pyproject.toml
2. **提交 `uv.lock`** - 确保团队依赖一致
3. **使用 `uv run`** - 自动激活虚拟环境
4. **配置镜像** - 加速国内下载
5. **定期清理缓存** - `uv cache clean`
