# uv 包管理器全面教程

uv 是 Astral 开发的极速 Python 包管理器，用 Rust 编写，比 pip 快 10-100 倍


## 1. 安装 uv


```python
# Arch/Manjaro 安装
!sudo pacman -S --needed --noconfirm uv
```

## 2. 配置国内镜像


```python
%%bash
export UV_DEFAULT_INDEX="https://pypi.tuna.tsinghua.edu.cn/simple"
```

## 3. 项目管理


```python
%%bash
DIR=/data/.manjaro/project-skills && mkdir -p $DIR
projectName=uv
# 初始化项目
cd $DIR && [[ -d $projectName ]] || uv init $projectName
# 创建虚拟环境
cd $DIR/$projectName && [[ -d .venv ]] || uv venv && uv add ipykernel pytest
```

```python
# 创建SKILLS.md
import pathlib
DIR = '/data/.manjaro/project-skills'
projectName = 'uv'
pathlib.Path(f'{DIR}/{projectName}/SKILLS.md').write_text(
    f'---\nname: {projectName}\ndescription: \n---\n# {projectName}\n'
)
```

```python
# 移除依赖
!uv remove requests
```

## 7. 运行命令


```python
# 在虚拟环境中运行脚本
!uv run python script.py
```

```python
# 运行模块
!uv run -m pytest
```

```python
# 运行测试
!uv run pytest
!uv run pytest tests/
!uv run pytest tests/test_example.py
```

```python
# 运行命令
!uv run black .
!uv run ruff check .
```

```python
# 安装全局工具 ipykernel
!uv tool install ruff
!uv tool install black
!uv tool install ipython
```

```python
# 列出已安装的工具
!uv tool list
```

```python
# 运行工具
!uv tool run ruff check .
```

```python
# 卸载工具
!uv tool uninstall ruff
```
