## 2. 基础示例：Hello World


```python
import gradio as gr

def greet(name):
    return f"Hello {name}!"

demo = gr.Interface(fn=greet, inputs="text", outputs="text")
demo.launch()
```

## 3. 常用输入组件


```python
import gradio as gr

def process(text, number, slider, checkbox, radio, dropdown):
    return f"文本: {text}, 数字: {number}, 滑块: {slider}, 复选框: {checkbox}, 单选: {radio}, 下拉: {dropdown}"

demo = gr.Interface(
    fn=process,
    inputs=[
        gr.Textbox(label="文本输入"),
        gr.Number(label="数字输入"),
        gr.Slider(0, 100, label="滑块"),
        gr.Checkbox(label="复选框"),
        gr.Radio(["选项1", "选项2"], label="单选"),
        gr.Dropdown(["A", "B", "C"], label="下拉菜单")
    ],
    outputs="text"
)
demo.launch()
```

## 4. 图像处理


```python
import gradio as gr
import numpy as np

def sepia(image):
    sepia_filter = np.array([[0.393, 0.769, 0.189],
                             [0.349, 0.686, 0.168],
                             [0.272, 0.534, 0.131]])
    sepia_img = image.dot(sepia_filter.T)
    sepia_img /= sepia_img.max()
    return sepia_img

demo = gr.Interface(fn=sepia, inputs="image", outputs="image")
demo.launch()
```

## 5. 多输入多输出


```python
import gradio as gr

def calculator(num1, operation, num2):
    if operation == "+":
        result = num1 + num2
    elif operation == "-":
        result = num1 - num2
    elif operation == "*":
        result = num1 * num2
    else:
        result = num1 / num2 if num2 != 0 else "错误"
    
    return result, f"{num1} {operation} {num2} = {result}"

demo = gr.Interface(
    fn=calculator,
    inputs=[
        gr.Number(label="数字1"),
        gr.Radio(["+", "-", "*", "/"], label="运算符"),
        gr.Number(label="数字2")
    ],
    outputs=[
        gr.Number(label="结果"),
        gr.Textbox(label="计算过程")
    ]
)
demo.launch()
```

## 6. Blocks：自定义布局


```python
import gradio as gr

def greet(name, intensity):
    return "Hello " + name + "!" * intensity

with gr.Blocks() as demo:
    gr.Markdown("# 欢迎使用 Gradio")
    
    with gr.Row():
        with gr.Column():
            name = gr.Textbox(label="姓名")
            intensity = gr.Slider(1, 10, value=5, label="强度")
            submit = gr.Button("提交")
        
        with gr.Column():
            output = gr.Textbox(label="输出")
    
    submit.click(fn=greet, inputs=[name, intensity], outputs=output)

demo.launch()
```

## 7. 聊天界面


```python
import gradio as gr

def chat(message, history):
    history = history or []
    response = f"你说: {message}"
    history.append((message, response))
    return history, history

with gr.Blocks() as demo:
    chatbot = gr.Chatbot()
    state = gr.State([])
    
    with gr.Row():
        txt = gr.Textbox(show_label=False, placeholder="输入消息...")
        submit = gr.Button("发送")
    
    submit.click(chat, [txt, state], [chatbot, state])
    txt.submit(chat, [txt, state], [chatbot, state])

demo.launch()
```

## 8. 数据可视化


```python
import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt

def plot_data(rows):
    df = pd.DataFrame({"x": range(rows), "y": range(rows)})
    
    fig, ax = plt.subplots()
    ax.plot(df["x"], df["y"])
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    
    return fig, df

demo = gr.Interface(
    fn=plot_data,
    inputs=gr.Slider(10, 100, value=50, label="数据点数量"),
    outputs=[gr.Plot(label="图表"), gr.Dataframe(label="数据")]
)
demo.launch()
```

## 9. 文件上传


```python
import gradio as gr

def process_file(file):
    with open(file.name, 'r') as f:
        content = f.read()
    return f"文件大小: {len(content)} 字符\n\n内容:\n{content[:500]}"

demo = gr.Interface(
    fn=process_file,
    inputs=gr.File(label="上传文件"),
    outputs="text"
)
demo.launch()
```

## 10. 实时更新


```python
import gradio as gr
import time

def slow_function(text):
    time.sleep(2)
    return text.upper()

demo = gr.Interface(
    fn=slow_function,
    inputs="text",
    outputs="text",
    live=True  # 实时更新
)
demo.launch()
```

## 11. TradingAgents + Gradio 示例


```python
import gradio as gr
from tradingagents.graph.trading_graph import TradingAgentsGraph
from tradingagents.default_config import DEFAULT_CONFIG

config = DEFAULT_CONFIG.copy()
config["deep_think_llm"] = "gpt-4o-mini"
config["quick_think_llm"] = "gpt-4o-mini"
config["max_debate_rounds"] = 1

ta = TradingAgentsGraph(debug=False, config=config)

def analyze_stock(symbol, date):
    try:
        _, decision = ta.propagate(symbol, date)
        return f"决策: {decision['action']}\n\n理由: {decision['reasoning']}"
    except Exception as e:
        return f"错误: {str(e)}"

with gr.Blocks() as demo:
    gr.Markdown("# TradingAgents 分析工具")
    
    with gr.Row():
        symbol = gr.Textbox(label="股票代码", value="AAPL")
        date = gr.Textbox(label="日期", value="2024-05-10")
    
    analyze_btn = gr.Button("分析")
    output = gr.Textbox(label="分析结果", lines=10)
    
    analyze_btn.click(fn=analyze_stock, inputs=[symbol, date], outputs=output)

demo.launch()
```

## 12. 高级功能：Tabs


```python
import gradio as gr

with gr.Blocks() as demo:
    with gr.Tabs():
        with gr.TabItem("文本处理"):
            text_input = gr.Textbox()
            text_output = gr.Textbox()
            text_button = gr.Button("转大写")
            text_button.click(lambda x: x.upper(), text_input, text_output)
        
        with gr.TabItem("计算器"):
            num1 = gr.Number()
            num2 = gr.Number()
            calc_output = gr.Number()
            calc_button = gr.Button("相加")
            calc_button.click(lambda x, y: x + y, [num1, num2], calc_output)

demo.launch()
```

## 13. 状态管理


```python
import gradio as gr

def increment(count):
    return count + 1, f"计数: {count + 1}"

with gr.Blocks() as demo:
    count_state = gr.State(0)
    output = gr.Textbox(label="输出")
    btn = gr.Button("点击计数")
    
    btn.click(increment, count_state, [count_state, output])

demo.launch()
```

## 15. 实践项目：股票分析仪表板


```python
import gradio as gr
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

def analyze_stock(symbol, period):
    stock = yf.Ticker(symbol)
    hist = stock.history(period=period)
    
    # 绘图
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(hist.index, hist['Close'])
    ax.set_title(f"{symbol} 股价")
    ax.set_xlabel("日期")
    ax.set_ylabel("价格")
    
    # 统计信息
    info = f"""当前价格: ${hist['Close'][-1]:.2f}
最高价: ${hist['High'].max():.2f}
最低价: ${hist['Low'].min():.2f}
平均价: ${hist['Close'].mean():.2f}"""
    
    return fig, info, hist.tail()

with gr.Blocks() as demo:
    gr.Markdown("# 股票分析仪表板")
    
    with gr.Row():
        symbol = gr.Textbox(label="股票代码", value="AAPL")
        period = gr.Dropdown(["1mo", "3mo", "6mo", "1y"], label="时间周期", value="3mo")
    
    analyze_btn = gr.Button("分析")
    
    with gr.Row():
        plot = gr.Plot(label="价格走势")
        stats = gr.Textbox(label="统计信息", lines=5)
    
    data_table = gr.Dataframe(label="最近数据")
    
    analyze_btn.click(analyze_stock, [symbol, period], [plot, stats, data_table])

demo.launch()
```
