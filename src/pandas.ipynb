## 1. 基础操作


```python
import pandas as pd
import numpy as np

# 创建 DataFrame
df = pd.DataFrame({
    'timestamp': [1, 2, 3, 4, 5],
    'open': [100, 102, 101, 103, 105],
    'high': [105, 106, 104, 107, 108],
    'low': [99, 101, 100, 102, 104],
    'close': [102, 101, 103, 105, 107],
    'volume': [1000, 1200, 1100, 1300, 1400]
})

print(df)
```

## 2. 数据选择


```python
# 选择列
print(df['close'])  # 单列
print(df[['open', 'close']])  # 多列

# 选择行
print(df.iloc[0])  # 第一行
print(df.iloc[-1])  # 最后一行
print(df.iloc[0:3])  # 前3行

# 选择特定值
print(df.iloc[0]['close'])  # 第一行的 close
print(df.loc[0, 'close'])  # 同上
```

## 3. 数据计算


```python
# 新增列
df['mid'] = (df['open'] + df['close']) / 2
df['range'] = df['high'] - df['low']

# 统计
print(df['close'].mean())  # 平均值
print(df['close'].max())   # 最大值
print(df['close'].min())   # 最小值
print(df['close'].std())   # 标准差

print(df)
```

## 4. 条件筛选


```python
# 单条件
print(df[df['close'] > 103])

# 多条件
print(df[(df['close'] > 102) & (df['volume'] > 1100)])

# 使用 query
print(df.query('close > 103 and volume > 1100'))
```

## 5. 滚动计算（技术指标）


```python
# 移动平均
df['ma5'] = df['close'].rolling(window=3).mean()

# 最大值/最小值
df['high_3'] = df['high'].rolling(window=3).max()
df['low_3'] = df['low'].rolling(window=3).min()

# 标准差
df['std_3'] = df['close'].rolling(window=3).std()

print(df)
```

## 6. 数据转换


```python
# 类型转换
df['close'] = df['close'].astype(float)

# 时间戳转换
df['datetime'] = pd.to_datetime(df['timestamp'], unit='ms')

# 四舍五入
df['close_round'] = df['close'].round(2)

print(df)
```

## 7. 实战：K线分析


```python
# 模拟 K 线数据
ohlcv = [
    [1770541380000, 70440.8, 70468.8, 70384.6, 70399.2, 13857.955],
    [1770541560000, 70399.2, 70453.2, 70386.3, 70424.2, 6125.307]
]

df = pd.DataFrame(ohlcv, columns=['timestamp', 'open', 'high', 'low', 'close', 'volume'])

# 计算中间价
df['mid'] = (df['open'] + df['close']) / 2

# 计算振幅
df['amplitude'] = ((df['high'] - df['low']) / df['low'] * 100).round(2)

# 涨跌幅
df['change'] = ((df['close'] - df['open']) / df['open'] * 100).round(2)

print(df)
```

## 8. 获取特定值


```python
# 获取第一根 K 线的最高价
high_price = df.iloc[0]['high']
print(f"最高价: {high_price}")

# 获取最后一根 K 线的收盘价
close_price = df.iloc[-1]['close']
print(f"收盘价: {close_price}")

# 获取所有收盘价的平均值
avg_close = df['close'].mean()
print(f"平均收盘价: {avg_close}")
```

## 9. 数据排序


```python
# 按收盘价排序
df_sorted = df.sort_values('close', ascending=False)
print(df_sorted)

# 按多列排序
df_sorted = df.sort_values(['volume', 'close'], ascending=[False, True])
print(df_sorted)
```

## 10. 数据合并


```python
# 纵向合并（追加行）
df1 = pd.DataFrame({'close': [100, 101]})
df2 = pd.DataFrame({'close': [102, 103]})
df_concat = pd.concat([df1, df2], ignore_index=True)
print(df_concat)

# 横向合并（追加列）
df3 = pd.DataFrame({'volume': [1000, 1100]})
df_merge = pd.concat([df1, df3], axis=1)
print(df_merge)
```
