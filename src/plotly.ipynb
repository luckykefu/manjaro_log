# Plotly 全面教程

本教程涵盖 Plotly 的主要功能和图表类型，包含实际示例。


## 1. 安装和导入


```python
# 安装 plotly
# !pip install plotly pandas numpy

import plotly.graph_objects as go
import plotly.express as px
import plotly.figure_factory as ff
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

## 2. 基础线图


```python
# 创建示例数据
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# 使用 graph_objects
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y1, mode='lines', name='sin(x)'))
fig.add_trace(go.Scatter(x=x, y=y2, mode='lines', name='cos(x)'))

fig.update_layout(
    title='三角函数图',
    xaxis_title='X轴',
    yaxis_title='Y轴'
)

fig.show()
```

## 3. 散点图


```python
# 创建随机数据
n = 100
df = pd.DataFrame({
    'x': np.random.randn(n),
    'y': np.random.randn(n),
    'size': np.random.randint(10, 50, n),
    'color': np.random.choice(['A', 'B', 'C'], n)
})

# 使用 plotly express
fig = px.scatter(df, x='x', y='y', size='size', color='color',
                 title='彩色气泡散点图',
                 hover_data=['size'])

fig.show()
```

## 4. 柱状图


```python
# 创建销售数据
categories = ['产品A', '产品B', '产品C', '产品D', '产品E']
q1_sales = [20, 14, 23, 25, 22]
q2_sales = [16, 18, 17, 19, 20]

fig = go.Figure(data=[
    go.Bar(name='Q1', x=categories, y=q1_sales),
    go.Bar(name='Q2', x=categories, y=q2_sales)
])

fig.update_layout(
    title='季度销售对比',
    barmode='group',
    xaxis_title='产品',
    yaxis_title='销售额 (万元)'
)

fig.show()
```

## 5. 饼图


```python
# 市场份额数据
labels = ['苹果', '三星', '华为', '小米', '其他']
values = [35, 25, 20, 10, 10]

fig = go.Figure(data=[go.Pie(labels=labels, values=values, hole=.3)])

fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(title='手机市场份额')

fig.show()
```

## 6. 热力图


```python
# 创建相关性矩阵
data = np.random.randn(10, 12)
months = ['1月', '2月', '3月', '4月', '5月', '6月',
          '7月', '8月', '9月', '10月', '11月', '12月']
cities = ['北京', '上海', '广州', '深圳', '杭州', '南京', '武汉', '成都', '西安', '重庆']

fig = go.Figure(data=go.Heatmap(
    z=data,
    x=months,
    y=cities,
    colorscale='Viridis'
))

fig.update_layout(title='城市月度温度热力图')
fig.show()
```

## 7. 3D 散点图


```python
# 3D 数据
n = 200
df_3d = pd.DataFrame({
    'x': np.random.randn(n),
    'y': np.random.randn(n),
    'z': np.random.randn(n),
    'color': np.random.choice(['类型1', '类型2', '类型3'], n)
})

fig = px.scatter_3d(df_3d, x='x', y='y', z='z', color='color',
                    title='3D 散点图')

fig.show()
```

## 8. 时间序列图


```python
# 创建时间序列数据
dates = pd.date_range('2023-01-01', periods=365, freq='D')
stock_price = 100 + np.cumsum(np.random.randn(365) * 0.5)
volume = np.random.randint(1000, 5000, 365)

# 创建子图
from plotly.subplots import make_subplots

fig = make_subplots(
    rows=2, cols=1,
    subplot_titles=('股价', '成交量'),
    vertical_spacing=0.1
)

fig.add_trace(
    go.Scatter(x=dates, y=stock_price, name='股价'),
    row=1, col=1
)

fig.add_trace(
    go.Bar(x=dates, y=volume, name='成交量'),
    row=2, col=1
)

fig.update_layout(title='股票价格和成交量', height=600)
fig.show()
```

## 9. 箱线图


```python
# 创建不同组的数据
np.random.seed(42)
group_a = np.random.normal(100, 15, 100)
group_b = np.random.normal(110, 20, 100)
group_c = np.random.normal(95, 10, 100)

fig = go.Figure()

fig.add_trace(go.Box(y=group_a, name='组A'))
fig.add_trace(go.Box(y=group_b, name='组B'))
fig.add_trace(go.Box(y=group_c, name='组C'))

fig.update_layout(title='不同组别数据分布箱线图')
fig.show()
```

## 10. 直方图


```python
# 正态分布数据
data1 = np.random.normal(100, 15, 1000)
data2 = np.random.normal(120, 20, 1000)

fig = go.Figure()

fig.add_trace(go.Histogram(x=data1, name='数据集1', opacity=0.7))
fig.add_trace(go.Histogram(x=data2, name='数据集2', opacity=0.7))

fig.update_layout(
    title='重叠直方图',
    barmode='overlay',
    xaxis_title='值',
    yaxis_title='频次'
)

fig.show()
```

## 11. 地理图


```python
# 城市数据
cities_data = {
    'city': ['北京', '上海', '广州', '深圳', '杭州'],
    'lat': [39.9042, 31.2304, 23.1291, 22.5431, 30.2741],
    'lon': [116.4074, 121.4737, 113.2644, 114.0579, 120.1551],
    'population': [2154, 2424, 1530, 1344, 981]
}

df_cities = pd.DataFrame(cities_data)

fig = px.scatter_mapbox(
    df_cities, 
    lat='lat', 
    lon='lon',
    size='population',
    hover_name='city',
    hover_data=['population'],
    title='中国主要城市人口分布'
)

fig.update_layout(
    mapbox_style='open-street-map',
    height=600
)

fig.show()
```

## 12. 动画图表


```python
# 创建动画数据
years = list(range(2010, 2021))
countries = ['中国', '美国', '日本', '德国', '英国']

data = []
for year in years:
    for country in countries:
        data.append({
            'year': year,
            'country': country,
            'gdp': np.random.randint(5, 20) + (year - 2010) * 0.5,
            'population': np.random.randint(50, 1400)
        })

df_anim = pd.DataFrame(data)

fig = px.scatter(
    df_anim, 
    x='gdp', 
    y='population',
    animation_frame='year',
    animation_group='country',
    size='gdp',
    color='country',
    hover_name='country',
    title='各国GDP与人口关系动画'
)

fig.show()
```

## 13. 仪表盘


```python
# 仪表盘图
fig = go.Figure(go.Indicator(
    mode = "gauge+number+delta",
    value = 420,
    domain = {'x': [0, 1], 'y': [0, 1]},
    title = {'text': "速度 (km/h)"},
    delta = {'reference': 380},
    gauge = {
        'axis': {'range': [None, 500]},
        'bar': {'color': "darkblue"},
        'steps': [
            {'range': [0, 250], 'color': "lightgray"},
            {'range': [250, 400], 'color': "gray"}
        ],
        'threshold': {
            'line': {'color': "red", 'width': 4},
            'thickness': 0.75,
            'value': 490
        }
    }
))

fig.show()
```

## 14. 瀑布图


```python
# 财务瀑布图
fig = go.Figure(go.Waterfall(
    name = "财务分析",
    orientation = "v",
    measure = ["relative", "relative", "relative", "relative", "total"],
    x = ["销售收入", "成本", "营销费用", "管理费用", "净利润"],
    textposition = "outside",
    text = ["+60", "-40", "-10", "-5", "5"],
    y = [60, -40, -10, -5, 0],
    connector = {"line":{"color":"rgb(63, 63, 63)"}},
))

fig.update_layout(title = "公司财务瀑布图")
fig.show()
```

## 15. 自定义样式和主题


```python
# 自定义主题
x = np.linspace(0, 10, 50)
y = np.sin(x)

fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=y, mode='lines+markers', name='数据'))

# 自定义布局
fig.update_layout(
    title={
        'text': '自定义样式图表',
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 24, 'color': 'darkblue'}
    },
    plot_bgcolor='lightgray',
    paper_bgcolor='white',
    font={'family': 'Arial', 'size': 14},
    xaxis={
        'title': 'X轴',
        'gridcolor': 'white',
        'linecolor': 'black'
    },
    yaxis={
        'title': 'Y轴',
        'gridcolor': 'white',
        'linecolor': 'black'
    }
)

fig.show()
```

## 16. 交互功能


```python
# 带下拉菜单的交互图表
x = np.linspace(0, 10, 100)

fig = go.Figure()

# 添加多条线
fig.add_trace(go.Scatter(x=x, y=np.sin(x), name='sin(x)', visible=True))
fig.add_trace(go.Scatter(x=x, y=np.cos(x), name='cos(x)', visible=False))
fig.add_trace(go.Scatter(x=x, y=np.tan(x), name='tan(x)', visible=False))

# 添加下拉菜单
fig.update_layout(
    updatemenus=[
        dict(
            buttons=list([
                dict(label="sin(x)",
                     method="update",
                     args=[{"visible": [True, False, False]}]),
                dict(label="cos(x)",
                     method="update",
                     args=[{"visible": [False, True, False]}]),
                dict(label="tan(x)",
                     method="update",
                     args=[{"visible": [False, False, True]}]),
                dict(label="全部",
                     method="update",
                     args=[{"visible": [True, True, True]}])
            ]),
            direction="down",
            showactive=True,
            x=0.1,
            xanchor="left",
            y=1.02,
            yanchor="top"
        ),
    ],
    title='交互式函数图表'
)

fig.show()
```

## 总结

本教程涵盖了 Plotly 的主要功能：

1. **基础图表**: 线图、散点图、柱状图、饼图
2. **高级图表**: 热力图、3D图、时间序列、箱线图
3. **特殊图表**: 地理图、动画图、仪表盘、瀑布图
4. **自定义**: 样式、主题、交互功能
5. **导出**: HTML、PNG、PDF格式

### 常用技巧：

- 使用 `px` (plotly express) 快速创建图表
- 使用 `go` (graph objects) 进行精细控制
- `update_layout()` 自定义图表外观
- `add_trace()` 添加多个数据系列
- 子图功能组合多个图表
