```python
%%bash
link_to_home() {
    local src_dir=${1:-/data/.home}
    for item in $src_dir/.* $src_dir/*; do
        local name=$(basename $item)
        [[ $name == '.' || $name == '..' ]] && continue
        local target=~/$name
        [[ -e $target && $(readlink -f $target) != $(readlink -f $item) ]] && rm -rf $target
        ln -sf $item ~/
    done
}

link_to_home
```
