```python
%%bash
set_mirror() {
    local url="${1:-https://mirrors.tuna.tsinghua.edu.cn/manjaro/stable/\$repo/\$arch}"
    tee /etc/pacman.d/mirrorlist <<EOF
# China mirrors tsinghua
Server = $url
EOF
}

# >>>>>>>>>>>>>>>>>>>>使用示例>>>>>>>>>>>>>>>
set_mirror "https://mirrors.tuna.tsinghua.edu.cn/manjaro/stable/\$repo/\$arch"
```
