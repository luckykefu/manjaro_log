# Amazon Q Developer 全面使用教程


## 一、核心功能

### 1. 代码生成

- 自然语言 → 代码
- 支持多种语言
- 上下文感知

### 2. 代码补全

- 实时建议
- 多行补全
- 智能预测

### 3. 代码解释

- 解释复杂代码
- 文档生成
- 注释添加

### 4. 代码优化

- 性能优化
- 安全扫描
- 最佳实践

### 5. 测试生成

- 单元测试
- 集成测试
- 测试用例


## 二、斜杠命令详解

### /dev - 开发新功能

```
用途: 实现新功能
示例: /dev 实现一个数据缓存功能

流程:
1. 分析需求
2. 设计方案
3. 生成代码
4. 创建文件
```

### /test - 生成测试

```
用途: 为代码生成测试
示例: /test 为 MaxPainFetcher 类生成单元测试

生成:
- 单元测试
- Mock 数据
- 边界测试
- 异常测试
```

### /review - 代码审查

```
用途: 审查代码质量
示例: /review @strategys/

检查:
- 代码规范
- 安全漏洞
- 性能问题
- 最佳实践
```

### /doc - 生成文档

```
用途: 生成文档
示例: /doc 为这个函数生成文档
```

### /fix - 修复问题

```
用途: 修复代码问题
示例: /fix 修复这个 bug
```

### /transform - 代码转换

```
用途: 转换代码
示例: /transform 将这段代码转换为 TypeScript
```


## 三、上下文引用

### @file - 引用文件

```
用法: @config.py
作用: 将文件内容加入上下文
```

### @folder - 引用文件夹

```
用法: @strategys/
作用: 将文件夹所有文件加入上下文
```

### @workspace - 引用工作区

```
用法: @workspace
作用: 智能选择相关文件
```

### @prompt - 引用提示词

```
用法: @prompt_name
作用: 使用保存的提示词模板
位置: ~/.aws/amazonq/prompts/
```


## 四、最佳实践

### 1. 清晰的需求描述

```
❌ 差: 写个函数
✅ 好: 写一个函数，接收 BTC 价格列表，计算 SMA(20)
```

### 2. 提供上下文

```
❌ 差: 优化这段代码
✅ 好: @config.py 优化配置加载性能
```

### 3. 分步骤执行

```
❌ 差: 实现完整的交易系统
✅ 好:
  1. /dev 实现数据采集模块
  2. /dev 实现策略计算模块
  3. /dev 实现订单执行模块
```

### 4. 验证结果

```
生成代码后:
1. /test 生成测试
2. /review 代码审查
3. 手动测试
```

## 六、高级技巧

### 1. 自定义提示词

```bash
# 创建提示词文件
mkdir -p ~/.aws/amazonq/prompts/
cat > ~/.aws/amazonq/prompts/trading_style.md << 'EOF'
# 交易代码风格
- 使用类型注解
- 添加日志记录
- 异常处理完善
- 遵循 PEP 8
EOF

# 使用
@trading_style /dev 实现策略
```

### 2. 项目级规则

```bash
# 创建规则文件
mkdir -p .amazonq/rules/
cat > .amazonq/rules/rules.md << 'EOF'
# 项目规则
- 使用中文注释
- 配置使用 YAML
- 日志使用 Rich
EOF

# 自动应用到所有对话
```

### 3. 多文件操作

```
# 同时引用多个文件
@config.py @utils.py @strategy.py
重构这些模块的依赖关系
```

### 4. 增量开发

```
# 第一步
/dev 实现基础框架

# 第二步
@framework.py /dev 添加错误处理

# 第三步
@framework.py /dev 添加日志记录

# 第四步
@framework.py /test 生成完整测试
```
