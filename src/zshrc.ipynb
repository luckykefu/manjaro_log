```python
%%bash
add_export() {
    local line="$1"
    local zshrc=${2:-~/.zshrc}
    grep -qF "$line" $zshrc || echo "$line" >> $zshrc
}

# uv
add_export 'export UV_DEFAULT_INDEX="https://pypi.tuna.tsinghua.edu.cn/simple"'
# appimage
add_export 'export PATH="$PATH:/data/.path/.appimage"'

# zsh-sudo plugin
add_export 'source /data/.zshplugins/zsh-sudo/zsh-sudo.zsh'

# proxy
add_export 'export ALL_PROXY="socks5://192.168.0.103:7897"'
```
