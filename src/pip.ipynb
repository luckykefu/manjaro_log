```python
%%bash
cfg_pip() {
    local url="${1:-https://pypi.tuna.tsinghua.edu.cn/simple/}"
    local dir="$HOME/.config/pip"
    
    mkdir -p "$dir"
    cat > "$dir/pip.conf" <<EOF
[global]
index-url = $url
trusted-host = $(echo "$url" | sed 's|https\?://||;s|/.*||')
timeout = 120
EOF
    echo "✓ pip config created"
}

# >>>>>>>>>>>>>>>>>>>>使用示例>>>>>>>>>>>>>>>
cfg_pip "https://pypi.tuna.tsinghua.edu.cn/simple/"
```
