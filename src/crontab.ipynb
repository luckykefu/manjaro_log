```python
%%bash
add_cron() {
    local job="$1"
    (crontab -l 2>/dev/null | grep -qF "$job") || (crontab -l 2>/dev/null; echo "$job") | crontab -
}

job="*/15 * * * * XDG_RUNTIME_DIR=/run/user/$(id -u) sh -c 'pactl set-sink-volume @DEFAULT_SINK@ 100% && paplay /usr/share/sounds/freedesktop/stereo/bell.oga'"
add_cron "$job"
```
