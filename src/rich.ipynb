## 五、Progress - 进度条


````python
from rich.progress import track
import time

# 简单进度条
for i in track(range(100), description="处理中..."):
    time.sleep(0.01)
````

````python
from rich.progress import Progress, SpinnerColumn, BarColumn, TextColumn

# 自定义进度条
with Progress(
    SpinnerColumn(),
    TextColumn("[progress.description]{task.description}"),
    BarColumn(),
    TextColumn("[progress.percentage]{task.percentage:>3.0f}%"),
) as progress:
    
    task1 = progress.add_task("[red]下载数据...", total=100)
    task2 = progress.add_task("[green]处理数据...", total=100)
    task3 = progress.add_task("[cyan]保存数据...", total=100)
    
    while not progress.finished:
        progress.update(task1, advance=0.5)
        progress.update(task2, advance=0.3)
        progress.update(task3, advance=0.2)
        time.sleep(0.02)
````

## 八、Syntax - 代码高亮


````python
from rich.syntax import Syntax
from rich.console import Console

console = Console()

code = '''
def calculate_sma(prices, period=20):
    """计算简单移动平均"""
    return sum(prices[-period:]) / period

# 使用
prices = [50000, 51000, 49000, 52000]
sma = calculate_sma(prices)
print(f"SMA: {sma}")
'''

syntax = Syntax(code, "python", theme="monokai", line_numbers=True)
console.print(syntax)
````

## 十、Live - 实时更新


````python
from rich.live import Live
from rich.table import Table
import time
import random

def generate_table():
    """生成实时价格表"""
    table = Table(title="实时行情")
    table.add_column("币种")
    table.add_column("价格")
    table.add_column("涨跌")
    
    btc_price = 50000 + random.randint(-1000, 1000)
    eth_price = 3000 + random.randint(-100, 100)
    
    table.add_row("BTC", f"${btc_price:,}", f"{random.uniform(-2, 2):.2f}%")
    table.add_row("ETH", f"${eth_price:,}", f"{random.uniform(-2, 2):.2f}%")
    
    return table

# 实时更新
with Live(generate_table(), refresh_per_second=1) as live:
    for _ in range(10):
        time.sleep(1)
        live.update(generate_table())
````

## 十六、实战案例


````python
from rich.console import Console
from rich.table import Table
from rich.panel import Panel
from rich.layout import Layout
from rich.live import Live
import time
import random

console = Console()

def create_dashboard():
    """创建交易仪表盘"""
    layout = Layout()
    
    layout.split_column(
        Layout(name="header", size=3),
        Layout(name="body"),
        Layout(name="footer", size=3)
    )
    
    # 标题
    layout["header"].update(
        Panel("[bold cyan]加密货币交易系统[/bold cyan]", style="bold magenta")
    )
    
    # 主体 - 价格表
    table = Table(title="实时行情")
    table.add_column("币种", style="cyan")
    table.add_column("价格", style="magenta")
    table.add_column("涨跌", style="green")
    table.add_column("成交量", style="yellow")
    
    btc_price = 50000 + random.randint(-1000, 1000)
    eth_price = 3000 + random.randint(-100, 100)
    
    table.add_row(
        "BTC",
        f"${btc_price:,}",
        f"{random.uniform(-2, 2):.2f}%",
        f"{random.randint(1000, 5000)}M"
    )
    table.add_row(
        "ETH",
        f"${eth_price:,}",
        f"{random.uniform(-2, 2):.2f}%",
        f"{random.randint(500, 2000)}M"
    )
    
    layout["body"].update(table)
    
    # 底部状态
    layout["footer"].update(
        Panel(f"[green]系统运行中[/green] | 更新时间: {time.strftime('%H:%M:%S')}", style="bold green")
    )
    
    return layout

# 实时更新仪表盘
with Live(create_dashboard(), refresh_per_second=1) as live:
    for _ in range(10):
        time.sleep(1)
        live.update(create_dashboard())
````
