```python
%%bash
install_miniconda() {
    local install_dir="${1:-/data/.path/.miniconda3}"
    local conda_bin="$install_dir/bin/conda"
    
    if [[ -f "$install_dir/bin/conda" ]]; then
        echo "✓ Miniconda3 already installed"
        return
    fi
    
    local save_path="$HOME/Downloads"
    local url="https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh"
    local installer="Miniconda3-latest-Linux-x86_64.sh"
    
    cd "$save_path"
    aria2c -x 16 -s 16 "$url" -o "$installer"
    
    rm -rf "$install_dir"
    mkdir -p "$(dirname "$install_dir")"
    bash "$installer" -b -f -p "$install_dir"
    
    
    if [[ -f "$conda_bin" ]]; then
        echo "✓ Miniconda3 installed"
        "$conda_bin" init zsh
        "$conda_bin" init bash
        "$conda_bin" config --set auto_activate_base false
        "$conda_bin" install ipykernel -y
        echo "✓ Miniconda3 configured"
    else
        echo "✗ Installation failed"
    fi
}

# >>>>>>>>>>>>>>>>>>>>使用示例>>>>>>>>>>>>>>>
install_miniconda "/data/.path/.miniconda3"
```
