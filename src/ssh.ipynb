```python
%%bash
gen_ssh_key() {
    local email=${1:-19157521820@163.com}
    local ssh_dir=${2:-/data/.home/.ssh}
    mkdir -p $ssh_dir
    [[ ! -f $ssh_dir/id_ed25519 ]] && ssh-keygen -t ed25519 -C $email -f $ssh_dir/id_ed25519 -N ""
    chmod 700 $ssh_dir && chmod 600 $ssh_dir/id_* && chmod 644 $ssh_dir/*.pub
    ln -sf $ssh_dir ~/.ssh
}

gen_ssh_key
```
