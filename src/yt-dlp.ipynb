### 代理配置


```python
# 设置代理环境变量（当前 session 生效）
import os
PROXY = "socks5h://192.168.0.103:7897"
os.environ["ALL_PROXY"] = PROXY
```

### 查看可用格式


```python
URL = "https://www.youtube.com/watch?v=ifWUYXnWJtY&list=PL_eZwOz1U5NlNgitPnn5synq-r6nXbw4l"
!yt-dlp --proxy {PROXY} \
    --cookies-from-browser firefox \
    -f "bestvideo+bestaudio" --merge-output-format mp4 \
    -o "~/Downloads/%(playlist_title)s/%(playlist_index)s - %(title)s.%(ext)s" \
    "{URL}"
```

### 下载视频


```python
# 最佳画质（视频+音频自动合并，/best 作为 fallback）
!yt-dlp --proxy {PROXY} -f "bestvideo+bestaudio/best" --merge-output-format mp4 \
    -o "~/Downloads/%(title)s.%(ext)s" "{URL}"
```

```python
# 指定分辨率（1080p）
!yt-dlp --proxy {PROXY} -f "bestvideo[height<=1080]+bestaudio/best[height<=1080]" --merge-output-format mp4 \
    -o "~/Downloads/%(title)s.%(ext)s" "{URL}"
```

### 仅下载音频


```python
# 提取为 mp3
!yt-dlp --proxy {PROXY} -x --audio-format mp3 --audio-quality 0 \
    -o "~/Downloads/%(title)s.%(ext)s" "{URL}"
```

### 下载字幕


```python
# 下载自动生成字幕（中文/英文）
!yt-dlp --proxy {PROXY} --write-auto-sub --sub-lang "zh-Hans,en" --skip-download \
    -o "~/Downloads/%(title)s.%(ext)s" "{URL}"
```

```python
# 视频+字幕一起下载
!yt-dlp --proxy {PROXY} -f "bestvideo+bestaudio" --merge-output-format mp4 \
    --write-auto-sub --sub-lang "zh-Hans,en" --embed-subs \
    -o "~/Downloads/%(title)s.%(ext)s" "{URL}"
```

### 下载播放列表


```python
PLAYLIST_URL = "https://www.youtube.com/playlist?list=PLAYLIST_ID"

# 下载整个播放列表
!yt-dlp --proxy {PROXY} -f "bestvideo+bestaudio/best" --merge-output-format mp4 \
    -o "~/Downloads/%(playlist_title)s/%(playlist_index)s - %(title)s.%(ext)s" \
    "{PLAYLIST_URL}"
```

```python
# 下载播放列表第 1-5 个视频
!yt-dlp --proxy {PROXY} --playlist-items 1-5 \
    -o "~/Downloads/%(playlist_index)s - %(title)s.%(ext)s" "{PLAYLIST_URL}"
```

### 断点续传 & 限速


```python
# 断点续传 + 限速 2MB/s
!yt-dlp --proxy {PROXY} -c --limit-rate 2M \
    -f "bestvideo+bestaudio" --merge-output-format mp4 \
    -o "~/Downloads/%(title)s.%(ext)s" "{URL}"
```

### 常用参数速查

| 参数 | 说明 |
|------|------|
| `-f FORMAT` | 指定格式，`bestvideo+bestaudio` 最佳画质 |
| `-x` | 仅提取音频 |
| `--audio-format mp3` | 音频转换格式 |
| `-o TEMPLATE` | 输出文件名模板 |
| `-c` | 断点续传 |
| `--proxy URL` | 设置代理 |
| `-F` | 列出所有可用格式 |
| `--write-auto-sub` | 下载自动字幕 |
| `--embed-subs` | 字幕嵌入视频 |
| `--playlist-items N` | 指定播放列表条目 |
| `--limit-rate RATE` | 限速，如 `2M` |
| `--cookies-from-browser chrome` | 使用浏览器 Cookie（会员内容）|


### Cookie 认证（解决 bot 检测）

**方案一**：直接读取浏览器 Cookie（需要 `python-secretstorage`）

**方案二**：导出 Cookie 文件（更稳定）
1. Chrome 安装扩展 [Get cookies.txt LOCALLY](https://chrome.google.com/webstore/detail/get-cookiestxt-locally/cclelndahbckbenkjhflpdbgdldlbecc)
2. 打开 YouTube 并登录，点击扩展导出 `cookies.txt` 到 `~/Downloads/`


```python
# 安装 secretstorage（方案一需要）
!sudo pacman -S --needed --noconfirm python-secretstorage
```

```python
# 方案一：直接读取 Chrome Cookie
!yt-dlp --proxy {PROXY} --cookies-from-browser chrome \
    -f "bestvideo+bestaudio/best" --merge-output-format mp4 \
    -o "~/Downloads/%(playlist_title)s/%(playlist_index)s - %(title)s.%(ext)s" "{URL}"
```

```python
# 方案二：使用导出的 Cookie 文件
COOKIES = "~/Downloads/cookies.txt"
!yt-dlp --proxy {PROXY} --cookies {COOKIES} \
    -f "bestvideo+bestaudio/best" --merge-output-format mp4 \
    -o "~/Downloads/%(playlist_title)s/%(playlist_index)s - %(title)s.%(ext)s" "{URL}"
```
