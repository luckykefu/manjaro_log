# path


```python
%%bash
add_to_path() {
    local tgt="${1:-/data/.path/.appimage}"
    local rc_file="${2:-$HOME/.zshrc}"
    
    mkdir -p "$tgt"
    local export_line="export PATH=\"$tgt:\$PATH\""
    
    if grep -qF "$export_line" "$rc_file"; then
        echo "✓ PATH already contains $tgt"
    else
        echo "$export_line" >>"$rc_file"
        echo "✓ Added $tgt to PATH"
    fi
}

# >>>>>>>>>>>>>>>>>>>>使用示例>>>>>>>>>>>>>>>
add_to_path "/data/.path/.appimage" "$HOME/.zshrc"
```
