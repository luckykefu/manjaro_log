# amazonq-mcp-installer

```python
%%bash
# [ 安装 amazonq-mcp-installer ]
MCP_DIR="$HOME/.aws/amazonq/MCP"
mkdir -p $MCP_DIR
cd $MCP_DIR

[[ -d amazonq-mcp-installer ]] || git clone https://github.com/bonjourzzz/amazonq-mcp-installer.git
cd amazonq-mcp-installer
npm install && npm run build
```

```python
%%bash
# [ 配置 Amazon Q default.json - 添加 amazonq-mcp-installer ]
CONFIG="$HOME/.aws/amazonq/agents/default.json"
MCP_PATH="$HOME/.aws/amazonq/MCP/amazonq-mcp-installer/build/index.js"

if jq -e '.mcpServers["amazonq-mcp-installer"]' $CONFIG > /dev/null 2>&1; then
    echo "已存在，跳过"
else
    jq --arg path "$MCP_PATH" \
        '.mcpServers["amazonq-mcp-installer"] = {"command": "node", "args": [$path]}' \
        $CONFIG > /tmp/default.json && mv /tmp/default.json $CONFIG
    echo "已添加"
fi
```
